# Explore restaurants

Read-only inspection of frozen application data and preserved audits. Run production construction with the package commands in the root README. Raw/interim evidence is separate from application inputs.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
root = Path.cwd().resolve()
if root.name == "notebooks":
    root = root.parent
data = pd.read_csv(root / "data/processed/restaurants.csv")
print("Canonical rows:", len(data))
display(data.head())
display(data.isna().sum().to_frame("missing"))
```

```python
display(data.price_category.value_counts(dropna=False).sort_index().to_frame("restaurants"))
display(data[["restaurant_id", "name", "price_category", "price_min", "price_max", "price_open_ended"]].head(10))
data.price_category.value_counts().sort_index().plot.bar(title="Experimental ordinal price categories")
plt.tight_layout()
plt.show()
```

```python
coverage = pd.read_csv(root / "data/reports/audit/reviews/google_local_review_coverage.csv")
display(coverage[coverage.entity_type.eq("restaurant")].head(10))
print("Missing review evidence means UNKNOWN, never a negative subjective label.")
```
