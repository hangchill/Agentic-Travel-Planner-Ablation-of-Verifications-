# Restaurant exploration
Inspect originals only. Final transformations belong in `scripts/preprocess_restaurants.py`.

```python
import pandas as pd
from pathlib import Path

root = Path.cwd().resolve()
if root.name == "notebooks":
    root = root.parent
raw_dir = root / "data" / "raw" / "restaurants"
```

```python
raw_files = sorted(p for p in raw_dir.rglob("*") if p.is_file())
for path in raw_files:
    print(path.relative_to(raw_dir))
csv_files = [p for p in raw_files if p.suffix.lower() == ".csv"]
```

```python
# Set this to a relative CSV path after examining the file list.
selected_csv = None
if selected_csv is not None:
    path = (raw_dir / selected_csv).resolve()
    if path not in [p.resolve() for p in csv_files]:
        raise ValueError("Select a CSV inside the raw restaurant directory")
elif len(csv_files) == 1:
    path = csv_files[0]
else:
    raise ValueError("Download data, then choose selected_csv explicitly if needed")
df = pd.read_csv(path)
```
