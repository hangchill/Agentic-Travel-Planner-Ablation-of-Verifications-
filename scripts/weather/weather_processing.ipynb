```python
import pandas as pd

# Fixed paths using clean forward slashes and corrected 'Manhattan' spelling
bronx = pd.read_csv('C:/NUS WORK/Y4S1/DSA4213 NLP/ProjectCode/data/raw/Bronx all year.csv')
brooklyn = pd.read_csv('C:/NUS WORK/Y4S1/DSA4213 NLP/ProjectCode/data/raw/Brooklyn all year.csv')
manhattan = pd.read_csv('C:/NUS WORK/Y4S1/DSA4213 NLP/ProjectCode/data/raw/Manhattan all year.csv')
queens = pd.read_csv('C:/NUS WORK/Y4S1/DSA4213 NLP/ProjectCode/data/raw/Queens all year.csv')
staten = pd.read_csv('C:/NUS WORK/Y4S1/DSA4213 NLP/ProjectCode/data/raw/Staten all year.csv')

bronx["area"] = "Bronx"
brooklyn["area"] = "Brooklyn"
manhattan["area"] = "Manhattan"
queens["area"] = "Queens"
staten["area"] = "Staten Island"

data = pd.concat([bronx, brooklyn, manhattan, queens, staten])

data.drop(columns = ['weather_code (wmo code)'], inplace=True)

data.info( )
```

```python
dict = {"time" : "date", 
        "temperature_2m_max (°C)" : "temperature_max (°C)", 
        "temperature_2m_min (°C)" : "temperature_min (°C)", 
        "rain_sum (mm)" : "rainfall (mm)", 
        "precipitation_hours (cm)" : "precipitation_hours", 
        "wind_speed_10m_max" : "wind_speed"}


data.rename(columns=dict, inplace=True)

data['date'] = pd.to_datetime(data['date'])

# The four seasonal 14-day windows (4 x 14 x 5 areas = 280 rows). A date can only fall in
# one window, so they must be combined with | (and), not &. Kept in `seasonal` for later
# tooling; `data` (the full year) is still what gets written to weather.csv below.
seasonal = data.loc[lambda df: 
    df['date'].between('2025-05-08', '2025-05-21') |  # Spring ; 14 days each season 
    df['date'].between('2025-06-08', '2025-06-21') |  # Summer 
    df['date'].between('2025-10-08', '2025-10-21') |  # Autumn 
    df['date'].between('2025-12-08', '2025-12-21')   # Winter 
]
print(seasonal.shape)

seasonal.info()
```

```python
seasonal.to_csv('C:/NUS WORK/Y4S1/DSA4213 NLP/ProjectCode/data/processed/weather_sample.csv')
```
